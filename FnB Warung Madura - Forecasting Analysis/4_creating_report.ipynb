# **Warung Madura Sales Forecasting**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
```

**Model Prediction**

```python
lstm_prediction_dir = 'result/data/lstm_prediction_results_reporting.csv'
lstm_prediction = pd.read_csv(lstm_prediction_dir)

xgboost_prediction_dir = 'result/data/xgboost_prediction_results_reporting.csv'
xgboost_prediction = pd.read_csv(xgboost_prediction_dir)

all_model_prediction_report = pd.concat([lstm_prediction, xgboost_prediction])
all_model_prediction_report.to_csv('result/data/prediction_results_reporting.csv')
```

**Model Performance Reporting**

```python
lstm_performance_dir = 'result/data/lstm_performance.csv'
lstm_performance = pd.read_csv(lstm_performance_dir)

xgboost_performance_dir = 'result/data/xgboost_performance.csv'
xgboost_performance = pd.read_csv(xgboost_performance_dir)

all_model_performance_report = pd.concat([lstm_performance, xgboost_performance])
all_model_performance_report.to_csv('result/data/model_performance.csv')
```
